```python
import random

class CPULookupLLM:
    def __init__(self, vocabulary, context_window=2):
        self.vocab = vocabulary
        self.context_window = context_window
        # In-memory lookup table replacing dense layers and matrix weights
        # Format: state_tuple -> {token: score}
        self.memory_table = {}

    def _get_state(self, context_tokens):
        """Creates a fixed-width state key from the last N tokens."""
        history = context_tokens[-self.context_window:]
        # Pad with a default token if history is too short
        while len(history) < self.context_window:
            history = ["<PAD>"] + history
        return tuple(history)

    def generate_next_token(self, context_tokens):
        """Performs a fast dictionary/index lookup instead of matrix multiplication."""
        state = self._get_state(context_tokens)

        # Initialize state if it's new
        if state not in self.memory_table:
            self.memory_table[state] = {token: 1.0 for token in self.vocab}

        action_scores = self.memory_table[state]

        # Vectorized-style discrete choice based on current scores
        tokens_list = list(action_scores.keys())
        scores = list(action_scores.values())

        # Prevent negative weights for random choice
        min_score = min(scores)
        if min_score < 0:
            scores = [s - min_score + 0.1 for s in scores]

        chosen_token = random.choices(tokens_list, weights=scores, k=1)[0]
        return chosen_token

    def reinforcement_update(self, context_tokens, chosen_token, reward, learning_rate=0.2):
        """Replaces gradient descent with discrete reinforcement learning updates."""
        state = self._get_state(context_tokens)
        if state not in self.memory_table:
            self.memory_table[state] = {token: 1.0 for token in self.vocab}

        # Update the score of the chosen token based on reward feedback
        old_score = self.memory_table[state][chosen_token]
        self.memory_table[state][chosen_token] = old_sore_updated = old_score + (learning_rate * reward)

# --- Example Testing & Training Script ---
if __name__ == "__main__":
    # Define a mini vocabulary for coding/text
    vocab = ["print", "(", "'Hello'", ")", "x", "=", "1", "+", "2", "<PAD>"]

    model = CPULookupLLM(vocab, context_window=2)

    print("--- Training via Reinforcement Learning Loop ---")
    target_token = "1" # Let's teach the model that after "x", "=" should be followed by "1"

    for episode in range(500):
        context = ["x", "="]
        next_token = model.generate_next_token(context)

        # Reward function: Positive reward for target, penalty otherwise
        if next_token == target_token:
            reward = 1.0
        else:
            reward = -0.5

        # Update lookup table directly without backpropagation
        model.reinforcement_update(context, next_token, reward)

    print("\n--- Testing Generation After Training ---")
    test_context = ["x", "="]
    predictions = [model.generate_next_token(test_context) for _ in range(5)]
    print(f"Context: {test_context}")
    print(f"Generated predictions over 5 trials: {predictions}")
```

```python
class TrieNode:
    def __init__(self):
        self.children = {}  # token -> TrieNode
        self.scores = {}    # token -> integer score

class ScalableCPULLM:
    def __init__(self, default_score=10):
        self.root = TrieNode()
        self.default_score = default_score

    def _traverse(self, context_tokens, create=False):
        """Navigates the Trie, optionally creating nodes if training."""
        curr = self.root
        for token in context_tokens:
            if token not in curr.children:
                if create:
                    curr.children[token] = TrieNode()
                else:
                    return None
            curr = curr.children[token]
        return curr

    def generate_next_token(self, context_tokens):
        """O(K) lookup where K is the context length, highly cache-efficient."""
        node = self._traverse(context_tokens, create=False)

        if not node or not node.scores:
            # Fallback uniform choice if context is unseen
            return "<UNK>"

        tokens = list(node.scores.keys())
        scores = list(node.scores.values())

        # Shift scores to be non-negative for random choice weighting
        min_s = min(scores)
        if min_s < 0:
            scores = [s - min_s + 1 for s in scores]

        import random
        return random.choices(tokens, weights=scores, k=1)[0]

    def reinforcement_update(self, context_tokens, chosen_token, reward, lr=1):
        """Updates scores in the Trie based on discrete rewards."""
        node = self._traverse(context_tokens, create=True)

        if chosen_token not in node.scores:
            node.scores[chosen_token] = self.default_score

        # Integer-based reward adjustment for CPU speed
        node.scores[chosen_token] += int(reward * lr)

# --- Testing the Scalable Version ---
if __name__ == "__main__":
    model = ScalableCPULLM()

    # Training sequence: simulating syntax learning
    training_data = [
        (["def", "func"], "("),
        (["def", "func"], "x"),
        (["if", "x"], ":"),
        (["if", "y"], ":")
    ]

    print("--- Training Scalable Trie Model ---")
    for _ in range(200):
        for context, target in training_data:
            prediction = model.generate_next_token(context)
            reward = 1 if prediction == target else -1
            model.reinforcement_update(context, target, reward)

    print("\n--- Testing Trie Generation ---")
    print(f"Context ['def', 'func'] -> Prediction: {model.generate_next_token(['def', 'func'])}")
    print(f"Context ['if', 'x'] -> Prediction: {model.generate_next_token(['if', 'x'])}")
```

```python
import pickle
import os

class PersistentCPULLM:
    def __init__(self, filepath="model_trie.bin", default_score=10):
        self.filepath = filepath
        self.default_score = default_score
        self.root = self._load_model()

    def _load_model(self):
        """Loads the Trie from disk if it exists, otherwise initializes a new root."""
        if os.path.exists(self.filepath):
            with open(self.filepath, "rb") as f:
                print(f"[Info] Loading model weights from {self.filepath}...")
                return pickle.load(f)
        print("[Info] Initializing a new CPU Trie model...")
        return {}

    def save_model(self):
        """Persists the Trie structure directly to disk for instant zero-copy access."""
        with open(self.filepath, "wb") as f:
            pickle.dump(self.root, f, protocol=pickle.HIGHEST_PROTOCOL)
        print(f"[Info] Model successfully saved to {self.filepath}")

    def _traverse(self, context_tokens, create=False):
        curr = self.root
        for token in context_tokens:
            if token not in curr:
                if create:
                    curr[token] = {}
                else:
                    return None
            curr = curr[token]
        return curr

    def generate_next_token(self, context_tokens):
        """O(K) lookup optimized for CPU L1/L2 cache traversal."""
        node = self._traverse(context_tokens, create=False)
        if not node or "__scores__" not in node:
            return "<UNK>"

        scores_dict = node["__scores__"]
        tokens = list(scores_dict.keys())
        scores = list(scores_dict.values())

        # Non-negative shift for probability weighting
        min_s = min(scores)
        if min_s < 0:
            scores = [s - min_s + 1 for s in scores]

        import random
        return random.choices(tokens, weights=scores, k=1)[0]

    def reinforcement_update(self, context_tokens, chosen_token, reward, lr=1):
        """Updates discrete integer scores based on reward feedback."""
        node = self._traverse(context_tokens, create=True)
        if "__scores__" not in node:
            node["__scores__"] = {}

        if chosen_token not in node["__scores__"]:
            node["__scores__"][chosen_token] = self.default_score

        # Integer math for rapid CPU execution
        node["__scores__"][chosen_token] += int(reward * lr)

# --- Execution & Verification ---
if __name__ == "__main__":
    model = PersistentCPULLM("cpu_coder_model.bin")

    # Simulate training a coding syntax rule: after "for", expect "i"
    coding_context = ["for"]
    target_token = "i"

    print("\n--- Running Reinforcement Loop ---")
    for _ in range(100):
        pred = model.generate_next_token(coding_context)
        reward = 1 if pred == target_token else -1
        model.reinforcement_update(coding_context, target_token, reward)

    # Save weights to disk
    model.save_model()

    print("\n--- Testing Persistent Generation ---")
    print(f"Context: {coding_context} -> Predicted Next Token: {model.generate_next_token(coding_context)}")
```

```python
def generate_beam_search(self, context_tokens, max_length=5, beam_width=2):
        """
        Performs beam search traversal across the Trie to generate
        multi-token code sequences deterministically on a CPU.
        """
        # Beam format: (accumulated_score, current_context_list, current_trie_node)
        # Initialize beam with the starting context
        node = self._traverse(context_tokens, create=False)
        if not node:
            return context_tokens # Fallback if context doesn't exist

        initial_score = 0.0
        beams = [(initial_score, list(context_tokens), node)]

        for _ in range(max_length):
            all_candidates = []

            for score, tokens, current_node in beams:
                # If we hit a dead end or a node without score data, keep the current beam
                if not current_node or "__scores__" not in current_node:
                    all_candidates.append((score, tokens, current_node))
                    continue

                # Explore all possible next tokens at this Trie node
                for next_token, token_score in current_node["__scores__"].items():
                    new_score = score + token_score
                    new_tokens = tokens + [next_token]

                    # Navigate deeper into the child node if it exists
                    next_node = current_node.get(next_token, None)
                    all_candidates.append((new_score, new_tokens, next_node))

            if not all_candidates:
                break

            # Sort candidates by highest score and keep top 'beam_width'
            all_candidates.sort(key=lambda x: x[0], reverse=True)
            beams = all_candidates[:beam_width]

        # Return the best generated sequence path
        best_beam = max(beams, key=lambda x: x[0])
        return best_beam[1]
```

```python
import os
import re
import urllib.request
import time
import random

# --- Step 1: Download & Load Tiny Shakespeare ---
DATA_URL = "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"
DATA_FILE = "input.txt"

def get_tinyshakespeare_tokens():
    if not os.path.exists(DATA_FILE):
        print(f"[Info] Downloading Tiny Shakespeare dataset from {DATA_URL}...")
        urllib.request.urlretrieve(DATA_URL, DATA_FILE)

    with open(DATA_FILE, "r", encoding="utf-8") as f:
        text = f.read()

    print(f"[Info] Raw dataset length: {len(text):,} characters")

    # Tokenize into words, line breaks, and punctuation using Regex
    tokens = re.findall(r"\w+|[^\w\s]|\n", text)
    print(f"[Info] Tokenized into {len(tokens):,} discrete tokens")
    return tokens

# --- Step 2: Trie Node Definition ---
class TrieNode:
    __slots__ = ['children', 'scores']
    def __init__(self):
        self.children = {}  # token -> TrieNode
        self.scores = {}    # next_token -> integer score

# --- Step 3: CPU-Native Trie Model with Discrete RL & Beam Search ---
class CPUTrieShakespeareLLM:
    def __init__(self, context_window=3, default_score=10):
        self.root = TrieNode()
        self.context_window = context_window
        self.default_score = default_score

    def _traverse(self, context_tokens, create=False):
        """O(K) path lookup across the Trie (where K is context length)."""
        curr = self.root
        # Use a slice up to context_window length
        ctx = context_tokens[-self.context_window:]
        for token in ctx:
            if token not in curr.children:
                if create:
                    curr.children[token] = TrieNode()
                else:
                    return None
            curr = curr.children[token]
        return curr

    def reinforcement_update(self, context_tokens, target_token, reward, lr=2):
        """Discrete RL update replacing gradient descent."""
        node = self._traverse(context_tokens, create=True)
        if target_token not in node.scores:
            node.scores[target_token] = self.default_score

        # Fast integer arithmetic score adjustment
        node.scores[target_token] += int(reward * lr)

    def generate_beam_search(self, start_tokens, generate_len=30, beam_width=3):
        """Multi-token generation using Beam Search path evaluation."""
        current_context = list(start_tokens)

        for _ in range(generate_len):
            ctx = current_context[-self.context_window:]
            node = self._traverse(ctx, create=False)

            if not node or not node.scores:
                # Fallback to random common punctuation/space if path is unlearned
                next_token = "\n"
            else:
                # Sort token scores at current node and pick top candidate
                candidates = sorted(node.scores.items(), key=lambda x: x[1], reverse=True)
                top_beam = candidates[:beam_width]

                # Weighted selection from the top-scoring beam options
                tokens = [c[0] for c in top_beam]
                weights = [max(1, c[1]) for c in top_beam]
                next_token = random.choices(tokens, weights=weights, k=1)[0]

            current_context.append(next_token)

        return "".join(
            [f" {t}" if t.isalnum() and i > 0 and current_context[i-1].isalnum() else t
             for i, t in enumerate(current_context)]
        )

# --- Step 4: Training & Evaluation Execution ---
if __name__ == "__main__":
    # Load tokenized dataset
    tokens = get_tinyshakespeare_tokens()

    # Context window of 3 tokens (e.g., ["First", "Citizen", ":"])
    model = CPUTrieShakespeareLLM(context_window=3, default_score=10)

    # We will train on the first 100,000 tokens of Tiny Shakespeare
    train_slice = tokens[:100000]

    print("\n--- Training CPU Trie Model on Tiny Shakespeare ---")
    start_time = time.time()

    for i in range(len(train_slice) - 3):
        context = train_slice[i:i+3]
        target = train_slice[i+3]

        # Positive reward signal for valid token transitions present in the text
        model.reinforcement_update(context, target, reward=1)

        if (i + 1) % 25000 == 0:
            print(f"Processed {i + 1:,} token paths...")

    elapsed = time.time() - start_time
    print(f"\n[Success] Training complete in {elapsed:.2f} seconds on CPU!")
    print(f"Total Unique Paths Learned in Trie: {len(model.root.children):,} top-level branches")

    # --- Step 5: Test Generation with Classic Shakespeare Prompts ---
    print("\n--- Generating Shakespeare Text via Beam Search ---")

    prompts = [
        ["First", "Citizen", ":"],
        ["To", "be", ","],
        ["ROMEO", ":", "\n"]
    ]

    for prompt in prompts:
        print(f"\nPrompt: {''.join(prompt)}")
        print("Generated Output:")
        output_text = model.generate_beam_search(prompt, generate_len=40, beam_width=3)
        print("-" * 50)
        print(output_text)
        print("-" * 50)
```

```python
import re
import time
import random
import json
import sqlite3
from datasets import load_dataset

def tokenize_code(source_code):
    """Code-aware tokenizer preserving indentation, operators, and symbols."""
    pattern = r"(^[ \t]+|[=<>!+\-*/%&|^~]=|\->|\w+|[^\w\s])"
    tokens = re.findall(pattern, source_code, re.MULTILINE)
    return [t for t in tokens if t]

class SQLiteCPUNgramModel:
    def __init__(self, db_path="cpu_model.db", max_context_window=6, default_score=10):
        self.max_context_window = max_context_window
        self.default_score = default_score
        self.total_paths = 0

        # Initialize SQLite database
        self.conn = sqlite3.connect(db_path)
        self.cursor = self.conn.cursor()

        # Create table with an index on context for ultra-fast lookups
        self.cursor.execute("""
            CREATE TABLE IF NOT EXISTS ngrams (
                context TEXT PRIMARY KEY,
                scores TEXT
            )
        """)
        self.conn.commit()

        # Get initial path count if resuming
        self.cursor.execute("SELECT COUNT(*) FROM ngrams")
        self.total_paths = self.cursor.fetchone()[0]

    def _context_to_key(self, context_tokens):
        """Serialize context tokens into a unique lookup string."""
        return "||".join(context_tokens)

    def _traverse_with_backoff(self, context_tokens):
        """Backoff traversal from max window down to 1-gram using SQLite."""
        for window_size in range(min(len(context_tokens), self.max_context_window), 0, -1):
            ctx = context_tokens[-window_size:]
            key = self._context_to_key(ctx)

            self.cursor.execute("SELECT scores FROM ngrams WHERE context = ?", (key,))
            row = self.cursor.fetchone()
            if row:
                scores = json.loads(row[0])
                if scores:
                    return scores
        return None

    def reinforcement_update(self, context_tokens, target_token, reward=1, lr=2):
        ctx = context_tokens[-self.max_context_window:]
        key = self._context_to_key(ctx)

        # Fast SQLite fetch-update-upsert
        self.cursor.execute("SELECT scores FROM ngrams WHERE context = ?", (key,))
        row = self.cursor.fetchone()

        if row:
            scores = json.loads(row[0])
        else:
            scores = {}
            self.total_paths += 1

        if target_token not in scores:
            scores[target_token] = self.default_score
        scores[target_token] += int(reward * lr)

        self.cursor.execute("""
            INSERT OR REPLACE INTO ngrams (context, scores)
            VALUES (?, ?)
        """, (key, json.dumps(scores)))

    def commit(self):
        self.conn.commit()

    def generate_code_beam(self, prompt_text, generate_len=40, beam_width=3):
        current_context = tokenize_code(prompt_text)

        for _ in range(generate_len):
            scores = self._traverse_with_backoff(current_context)

            if not scores:
                next_token = "\n"
            else:
                candidates = sorted(scores.items(), key=lambda x: x[1], reverse=True)
                top_beam = candidates[:beam_width]
                tokens = [c[0] for c in top_beam]
                weights = [max(1, c[1]) for c in top_beam]
                next_token = random.choices(tokens, weights=weights, k=1)[0]

            current_context.append(next_token)

        # Reconstruct output preserving spacing
        output = []
        for i, t in enumerate(current_context):
            if i > 0 and t.isalnum() and current_context[i-1].isalnum():
                output.append(" " + t)
            else:
                output.append(t)
        return "".join(output)

# --- Scaled Execution with Disk Backing ---
if __name__ == "__main__":
    print("[Info] Connecting to FineWeb-Edu stream...")
    dataset = load_dataset("HuggingFaceFW/fineweb-edu", name="sample-10BT", split="train", streaming=True)

    model = SQLiteCPUNgramModel(db_path="code_assistant.db", max_context_window=6)

    TARGET_DOCS = 50000  # Now safely scalable to 50k+ docs without RAM issues!
    print(f"\n--- Starting Disk-Backed Scaled Training (Target: {TARGET_DOCS:,} Documents) ---")
    start_time = time.time()

    tokens_processed = 0
    doc_count = 0

    for item in dataset:
        if doc_count >= TARGET_DOCS:
            break

        text = item.get("text", "")
        if not text:
            continue

        tokens = tokenize_code(text)
        if len(tokens) < model.max_context_window + 1:
            continue

        tokens = tokens[:1200]
        for i in range(len(tokens) - model.max_context_window):
            context = tokens[i : i + model.max_context_window]
            target = tokens[i + model.max_context_window]
            model.reinforcement_update(context, target, reward=1)
            tokens_processed += 1

        doc_count += 1

        # Commit transactions periodically for speed and safety
        if doc_count % 1000 == 0:
            model.commit()
            elapsed = time.time() - start_time
            print(f"Processed {doc_count:,} docs | {tokens_processed:,} tokens | Unique Paths on Disk: {model.total_paths:,} | Time: {elapsed:.2f}s")

    model.commit()
    total_elapsed = time.time() - start_time
    print(f"\n[Success] Scaled training complete in {total_elapsed:.2f} seconds!")
    print(f"Total Unique Syntax Paths Indexed: {model.total_paths:,}")

    # --- Test Generation ---
    print("\n--- Testing Disk-Backed Scaled Generation ---")
    prompts = [
        "def calculate_sum(a,",
        "import os\ndef get_",
        "for i in range("
    ]

    for prompt_str in prompts:
        print(f"\nPrompt: {prompt_str}")
        print("Generated Output:")
        print("-" * 50)
        print(model.generate_code_beam(prompt_str, generate_len=35, beam_width=3))
        print("-" * 50)
```
